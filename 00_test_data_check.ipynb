```python
import pandas as pd

pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', None)

# 회사에서의 경로: r'C:\Users\20162180\AI_Data_exhibition\parqueted_data\test.parquet'

test = pd.read_parquet(r'C:\Users\20162180\AI_Data_exhibition\parqueted_data\test.parquet')
# test.head(10)
```

```python
# 강우사상 분리 후 개수가 데이터설명서(30개)와 같은지 확인: 1-3 노트북 참고

test_event_time = (
    test[['timestamp']]
    .drop_duplicates()
    .sort_values('timestamp')
    .reset_index(drop=True)
)

test_event_time['test_event_time_diff'] = (
    test_event_time['timestamp']
    .diff()
)

cond_test_split_time = (
    test_event_time['test_event_time_diff'] != pd.Timedelta(hours = 1)
)

test_event_time['test_event_id'] = cond_test_split_time.cumsum()

print(test_event_time['test_event_id'].nunique())
# test_event_time
```
