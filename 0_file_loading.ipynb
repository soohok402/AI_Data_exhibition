```python
import pandas as pd

# ============================================================
# 1. 원본 csv.gz 불러오기
# ============================================================
train = pd.read_csv(
    r'C:\Users\soohok402\내 드라이브\Github\AI_Data_exhibition\raw_data\train.csv.gz'
)

test = pd.read_csv(
    r'C:\Users\soohok402\내 드라이브\Github\AI_Data_exhibition\raw_data\test.csv.gz'
)


# ============================================================
# 2. timestamp를 datetime으로 변환
#    - 잘못된 날짜는 NaT(Not a Time: 시간결측치)로 처리
# ============================================================
train['timestamp'] = pd.to_datetime(
    train['timestamp'],
    errors='coerce'  # 날짜로 변환할 수 없는 값은 에러를 내지 않고 NaT로 반환
)

test['timestamp'] = pd.to_datetime(
    test['timestamp'],
    errors='coerce'
)


# ============================================================
# 3. timestamp 변환 오류 확인
# ============================================================
print('train timestamp 오류:', train['timestamp'].isnull().sum())
print('test timestamp 오류:', test['timestamp'].isnull().sum())


# ============================================================
# 4. row_id 고유성 확인
# ============================================================
print('train row_id unique:', train['row_id'].is_unique)
print('test row_id unique:', test['row_id'].is_unique)
# 고유성 확인됨(True)

# ============================================================
# 5. row_id를 인덱스로 지정
# ============================================================
train = train.set_index('row_id')
test = test.set_index('row_id')


# ============================================================
# 6. 분석용 parquet 파일로 저장
#    - row_id를 인덱스로 저장해야 하므로 index=False 사용하지 않음
# ============================================================
train.to_parquet(
    r'C:\Users\soohok402\내 드라이브\Github\AI_Data_exhibition\parqueted_data\train.parquet'
)

test.to_parquet(
    r'C:\Users\soohok402\내 드라이브\Github\AI_Data_exhibition\parqueted_data\test.parquet'
)


# ============================================================
# 7. 저장 완료 확인
# ============================================================
print('Parquet 저장 완료')
```
