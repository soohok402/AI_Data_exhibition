```python
from pathlib import Path
PROJECT_ROOT = Path.cwd().parent

import pandas as pd

pd.set_option('display.max_columns', None)  # none은 제한값 자리(몇개 열까지 보여줄까)
pd.set_option('display.max_rows', None)

# ============================================================
# 0. 데이터 불러오기
# ============================================================

train = pd.read_parquet(PROJECT_ROOT / 'parqueted_data' / 'train_split.parquet')
val = pd.read_parquet(PROJECT_ROOT / 'parqueted_data' / 'val_split.parquet')
```

```python
# ============================================================
# 1. 수위 차분 파생변수 생성
# ============================================================
# wl_diff_00h = wl_t(현재수위) - wl_t_minus_00h 파생변수 5개 생성

train_fe = train.copy()
val_fe = val.copy()

for h in [1, 3, 6, 12, 24]:
    train_fe[f"wl_diff_{h}h"] = (
        train['wl_t'] - train[f"wl_t_minus_{h}h"]
    )
    val_fe[f"wl_diff_{h}h"] = (
        val['wl_t'] - val[f"wl_t_minus_{h}h"]
    )

display(train_fe.head())
print()
display(val_fe.head())
```

```python
# ============================================================
# 2. 수위 차분 파생변수 생성
# ============================================================
# imerg_int_0to0h = (imerg_rain_00h - imerg_rain_00h)/hours
# sfc_int_0to0h = (sfc_rain_00h - sfc_rain_00h)/hours


for df in [train_fe, val_fe]:

    df['imerg_int_1to6h'] = (df['imerg_rain_6h'] - df['imerg_rain_1h']) / 5
    df['imerg_int_6to24h'] = (df['imerg_rain_24h'] - df['imerg_rain_6h']) / 18

    df['sfc_int_1to6h'] = (df['sfc_rain_6h'] - df['sfc_rain_1h']) / 5
    df['sfc_int_6to24h'] = (df['sfc_rain_24h'] - df['sfc_rain_6h']) / 18

display(train_fe.head())
print()
display(val_fe.head())
```

```python
# ============================================================
# 2. 데이터셋 저장하기
# ============================================================

train_fe.to_parquet(PROJECT_ROOT / 'parqueted_data' / '1_train_fe.parquet')
val_fe.to_parquet(PROJECT_ROOT / 'parqueted_data' / '1_val_fe.parquet')
```
