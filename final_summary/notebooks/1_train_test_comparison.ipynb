```python
from pathlib import Path
PROJECT_ROOT = Path.cwd().parent

import pandas as pd

# pd.set_option('display.max_columns', None)
# pd.set_option('display.max_rows', None)

# ============================================================
# 0. 데이터 불러오기
# ============================================================

train = pd.read_parquet(PROJECT_ROOT / 'data' / '1_train_full_fe.parquet')
test = pd.read_parquet(PROJECT_ROOT / 'data' / '1_test_fe.parquet')

print(train.shape, test.shape)
```

```python
# ============================================================
# 1. 컬럼군 정의
# ============================================================

# 종속변수군
target_col = 'target_maxrise_6h'

# 수위 컬럼군
wl_cols = [
    'wl_t',
    'wl_t_minus_1h',
    'wl_t_minus_3h',
    'wl_t_minus_6h',
    'wl_t_minus_12h',
    'wl_t_minus_24h',
]

# 수위차분(wl_diff) 컬럼군
wl_diff_cols = [
    'wl_t',
    'wl_diff_1h',
    'wl_diff_3h',
    'wl_diff_6h',
    'wl_diff_12h',
    'wl_diff_24h'
]

# 강수 컬럼군
rain_cols = [
    # IMERG강우
    'imerg_rain_1h',
    'imerg_rain_6h',
    'imerg_rain_24h',
    # 지상강우
    'sfc_rain_1h',
    'sfc_rain_6h',
    'sfc_rain_24h',
    # 레이더 반사도
    'radar_reflectivity_mean',
    'radar_reflectivity_p95',
    'radar_reflectivity_ge20_fraction'
]

# 강우강도 컬럼군
rain_int_cols = [
    'imerg_rain_1h',
    'imerg_int_1to6h',
    'imerg_int_6to24h',
    'sfc_rain_1h',
    'sfc_int_1to6h',
    'sfc_int_6to24h'
]

# 환경 컬럼군
env_cols = [
    'cat_area',       # km²
    'wsarea',         # km²
    'mean_elevation', # m
    'mean_slope',     # °
    'urban_ratio',    # %
    'forest_ratio',   # %
    'stream_order'
]

num_cols = train.select_dtypes(include='number').columns

assert len(num_cols) == 32, "'num_cols' 개수 이상"
assert (set(train.columns) - set(num_cols)) == {'station_id', 'timestamp'}, "'num_cols' 확인필요"
```

```python
# ============================================================
# 2. station_id의 유역환경변수 유일성 확인하기
# ============================================================
# train, test셋의 station_id가 동일하면 env_cols는 서로 동일

print("===== Test에만 존재하는 컬럼 (없어야 정상) =====")
print(set(test.columns) - set(train.columns))

print("===== Train에만 존재하는 컬럼 (종속변수만 있어야 정상) =====")
print(set(train.columns) - set(test.columns))

assert (train.groupby('station_id')[env_cols].nunique().max() == 1).all(),  "관측소별 유역환경변수군의 데이터가 동일하지 않음"

```

```python
# ======================================================================
# 3-1. train, test 데이터 박스플롯, 히스토그램 비교하기: 분포 비교용
# ======================================================================

import matplotlib.pyplot as plt

fig, axes = plt.subplots(
    2 * len(num_cols),
    2,
    figsize=(12, 8 * len(num_cols))
)

if len(num_cols) == 1:
    axes = [axes]

for i, col in enumerate(
    [col for col in num_cols if col not in [target_col]]
):

    tr = train[col].dropna()
    te = test[col].dropna()

    axes[i][0].boxplot(
        [tr, te],
        tick_labels=['train', 'Test']  # X축 이름
    )
    axes[i][0].set_title(f"{col}-Boxplot", fontsize=20)
    axes[i][0].tick_params(labelsize=15)

    axes[i][1].hist(
        tr,
        range=(min(tr.min(), te.min()), max(tr.max(), te.max())),
        bins=30,
        alpha=0.5,
        density=True,
        label='Train'
    )
    axes[i][1].hist(
        te,
        range=(min(tr.min(), te.min()), max(tr.max(), te.max())),
        bins=30,
        alpha=0.5,
        density=True,
        label='Test'
    )
    axes[i][1].set_title(f"{col}-Histogram(Log-scale)", fontsize=20)
    axes[i][1].tick_params(labelsize=15)
    axes[i][1].legend()  # 범례
    axes[i][1].set_yscale('log')  # 히스토그램, 극단값의 낮은 값도 잘 보이게 로그스케일로

plt.tight_layout()
plt.show()
```
