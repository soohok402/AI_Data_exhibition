```python
# ============================================================
# 실제값 - 예측잔차 산점도 (조건부 강조)
# ============================================================

# 1. 분석 설정
df = train_kf_oof          # 분석 데이터
x_col = 'y_kf_val'         # X축: 실제값
y_col = 'residual'         # Y축: 예측잔차
plot_title = 'Actual vs Residual'

# 2. 강조 조건 설정 (True = 전체 데이터)
cond = (
    (df['station_id'] == 101)
    # & (df['hour'] >= 12)
    # & (df['hour'] < 18)
)

# 3. 조건별 데이터 구분
highlight = df.loc[cond]   # 빨간색 강조 데이터

# 4. 산점도
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(10, 6))

# 전체 데이터: 회색
ax.scatter(
    df[x_col], df[y_col],
    color='gray', alpha=0.15, s=5,
    label='All data'
)

# 조건 데이터: 빨간색
ax.scatter(
    highlight[x_col], highlight[y_col],
    color='red', alpha=0.7, s=10,
    label=f'Highlighted (n={len(highlight):,})'
)

# 5. 그래프 설정
ax.axhline(0, color='black', linestyle='--', linewidth=1)
ax.set(
    title=plot_title,
    xlabel=x_col,
    ylabel=y_col
)
ax.legend()
ax.grid(alpha=0.2)

plt.tight_layout()
plt.show()

# 6. 강조 데이터 확인
print(f"전체 데이터: {len(df):,}")
print(f"강조 데이터: {len(highlight):,}")
print(f"강조 비율: {len(highlight)/len(df):.2%}")
```
