```python
# ============================================================
# 4. 시계열 구조 및 강우사상 경계 확인
# ============================================================
# - 관측소별 timestamp 정렬
# - 관측시점 간 시간간격(diff) 확인
# - 시간간격을 기준으로 event 간격 탐색, 강우사상별 event_id 생성

# 4-1. 관측소별 timestamp 정렬
time_check = (
    df.loc[:, ['station_id', 'timestamp']]  # 전부 다 보려고 하니 무거워서 에러남
      .sort_values(['station_id', 'timestamp'])
  )
# 4-2 위 df 옆에 차분된 결과(시간간격)를 첨가
time_check['time_diff'] = (
      time_check.groupby('station_id')['timestamp'].diff()  # groupby를 해줘야 한 station_id 내에서 diff()를 끝냄
)
display(time_check.head())

# 4-3 event_id 생성하기




#       .groupby('station_id')['timestamp']  # 관측소별로 그루핑, 시각 출력
#       .diff()  # 이전행과의 차이를 출력하는 함수, timestamp의 차이만 들어있는 시리즈가 출력
# 보기 불편,,

```
